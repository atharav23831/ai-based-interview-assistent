# AI Mock Interview - ML Model Training

This notebook demonstrates the pipeline to train a machine learning model to evaluate interview answers.

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
import joblib
import sys
sys.path.append('../src')

from preprocessing import preprocess_dataframe
from features import extract_features
```

## 1. Load Data

```python
df = pd.read_csv('../data/interview_dataset.csv')
df.head()
```

## 2. Preprocessing & Feature Extraction

```python
df = preprocess_dataframe(df)
X, vectorizer, df = extract_features(df, is_train=True)
print("Feature matrix shape:", X.shape)
```

## 3. Train Test Split

```python
y_class = df['quality']
y_score = df['score']

X_train, X_test, y_class_train, y_class_test, y_score_train, y_score_test = train_test_split(
    X, y_class, y_score, test_size=0.2, random_state=42
)
```

## 4. Train Model
We can refer to `src/train.py` for the full training script that tests multiple algorithms. Here we can train one for demonstration.

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report, accuracy_score

model = RandomForestClassifier(n_estimators=100, random_state=42)
model.fit(X_train, y_class_train)

y_pred = model.predict(X_test)
print("Accuracy:", accuracy_score(y_class_test, y_pred))
print("\nClassification Report:\n", classification_report(y_class_test, y_pred))
```
